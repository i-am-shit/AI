```python
import random

class VacuumSystem:
    def __init__(self):
        self.environment_state = "Dirty"

    def sensor_read_location_A(self):
        return self.environment_state

    def agent_brain(self, percept):
        if percept == "Dirty":
            return "Suck"
        else:
            return "NoOp"

    def actuator_execute(self, action):
        if action == "Suck":
            self.environment_state = "Clean"
            print("       Actuator: Vacuum motor activated. (Room cleaned)")
        elif action == "NoOp":
            print("       Actuator: Remained idle. (Power saving)")

    def environment_dynamic_change(self):
        if random.random() < 0.40:
            self.environment_state = "Dirty"
            print("       Environment Change: Dirt randomly fell on the floor!")


def run_simulation():
    robot = VacuumSystem()
    print("Vacuum Robot System Initiated (Location A)")

    for step in range(1, 5):
        print(f"\n[Step {step}]")

        current_percept = robot.sensor_read_location_A()
        print(f"       Sensor Read: Location A is {current_percept}")

        chosen_action = robot.agent_brain(current_percept)
        print(f"       Agent Decision: Chosen action -> {chosen_action}")

        robot.actuator_execute(chosen_action)

        if step < 4:
            robot.environment_dynamic_change()



if __name__ == "__main__":
    run_simulation()
print("AMSHITH 1WN24CS035")
```
